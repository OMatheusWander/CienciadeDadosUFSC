```python
def escalonamento_logica_bruta(A, b):
    # 1. Criar matriz aumentada manualmente
    n = len(A)
    cols = len(A[0])
    matriz = []
    for i in range(n):
        linha = []
        for j in range(cols):
            linha.append(float(A[i][j]))
        linha.append(float(b[i]))
        matriz.append(linha)

    print("Matriz Aumentada Inicial:")
    for l in matriz: print(l)
    print("-" * 30)

    # Gauss-Jordan
    for i in range(n):
        # Encontrar pivô (Lógica bruta de busca)
        max_val = abs(matriz[i][i])
        max_row = i
        for k in range(i + 1, n):
            if abs(matriz[k][i]) > max_val:
                max_val = abs(matriz[k][i])
                max_row = k

        # Troca de linhas manual
        if max_row != i:
            matriz[i], matriz[max_row] = matriz[max_row], matriz[i]
            print(f"Passo: Troca L{i} <-> L{max_row}")
            for l in matriz: print(l)

        # Normalização do pivô
        pivo = matriz[i][i]
        if pivo != 0:
            for j in range(i, cols + 1):
                matriz[i][j] /= pivo
            print(f"Passo: L{i} = L{i} / {pivo:.2f}")
            for l in matriz: print(l)

        # Eliminação (Zerar coluna)
        for k in range(n):
            if k != i:
                fator = matriz[k][i]
                if fator != 0:
                    for j in range(i, cols + 1):
                        matriz[k][j] = matriz[k][j] - (fator * matriz[i][j])
                    print(f"Passo: L{k} = L{k} - ({fator:.2f}) * L{i}")
                    for l in matriz: print(l)
        print("-" * 30)

    return matriz

# Dados de teste
A_teste = [[3, -1, 2], [1, 2, 3], [2, -2, 2]]
B_teste = [12, 11, 2]

resultado = escalonamento_logica_bruta(A_teste, B_teste)
```
