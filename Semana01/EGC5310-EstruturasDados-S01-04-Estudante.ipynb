## 4. Tempo × número de operações

Agora mediremos o tempo para observar que **tempo de execução** e **ordem de crescimento** não são a mesma coisa.


```python
def medir_busca(dados, matricula, repeticoes=7):
    tempos = []

    for _ in range(repeticoes):
        inicio = perf_counter()
        buscar(dados, matricula)
        tempos.append(perf_counter() - inicio)

    return median(tempos)
```

```python
dados = gerar_estudantes(100_000)

for i in range(5):
    tempo = medir_busca(dados, -1, repeticoes=1)
    print(f"execução {i + 1}: {tempo:.8f} s")
```
