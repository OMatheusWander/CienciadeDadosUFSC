```python
#Exercício 8
import random

# Exercício 8
nomes_produtos = [
    'Notebook', 'Caderno', 'Jaqueta', 'Açúcar', 'Teclado',
    'Boné', 'Calça', 'Shampoo', 'Camiseta', 'Cadeira',
    'Mochila', 'Borracha', 'Impressora', 'Mesa', 'Monitor',
    'Sabonete', 'Macarrão', 'Desodorante', 'Mouse', 'Café',
    'Tênis', 'Arroz', 'Luminária', 'Armário', 'Gaveteiro',
    'Pasta de Dente', 'Escova', 'Caneta', 'Feijão', 'Lápis'
]

categorias = [
    'Informática', 'Móveis', 'Papelaria',
    'Vestuário', 'Alimentos', 'Higiene'
]

prod_estoque_maior_20 = []
prod_preco_maior_100 = []
categorias_agrupadas = {}  # Alterado para dicionário 🗂️

print('VALOR TOTAL EM ESTOQUE DE CADA PRODUTO')
print(80*'-')

# Loop principal pelos produtos
for produto in nomes_produtos:
    print(f'\nProduto: {produto}')
    print('Escolha uma categoria:')
    print('1 - Informática\n2 - Móveis\n3 - Papelaria\n4 - Vestuário\n5 - Alimentos\n6 - Higiene')

    input_categoria = input('Digite o número da categoria: ')

    if input_categoria == '1':
        categoria = 'Informática'
    elif input_categoria == '2':
        categoria = 'Móveis'
    elif input_categoria == '3':
        categoria = 'Papelaria'
    elif input_categoria == '4':
        categoria = 'Vestuário'
    elif input_categoria == '5':
        categoria = 'Alimentos'
    elif input_categoria == '6':
        categoria = 'Higiene'
    else:
        categoria = 'Não Informada'

    # Geração dos dados aleatórios 🎲
    estoque = random.randint(0, 50)
    preco = round(random.uniform(1.00, 300.00), 2)

    # Criação do dicionário do produto
    dic_produtos = {
        'nome': produto,
        'categoria': categoria,
        'dados_estoque_preco': [estoque, preco]
    }

    # Cálculo e exibição do valor total em estoque 💰
    valor_estoque = estoque * preco
    print(f'-> Produto: {produto} | Total em Estoque: R$ {valor_estoque:.2f}')

    # Filtros de estoque e preço 🔍
    if dic_produtos['dados_estoque_preco'][0] > 20:
        prod_estoque_maior_20.append(dic_produtos)

    if dic_produtos['dados_estoque_preco'][1] > 100:
        prod_preco_maior_100.append(dic_produtos)

    # Agrupamento por categoria 🏷️
    if categoria not in categorias_agrupadas:
        categorias_agrupadas[categoria] = []
    categorias_agrupadas[categoria].append(dic_produtos)

# --- Relatórios Finais ---

print('\nPRODUTOS COM ESTOQUE MAIOR QUE 20')
print(80*'-')
print(prod_estoque_maior_20)

print('\nPRODUTOS COM PREÇO ACIMA DE R$100')
print(80*'-')
print(prod_preco_maior_100)

print('\nPRODUTOS AGRUPADOS POR CATEGORIA')
print(80*'-')
for cat, prods in categorias_agrupadas.items():
    print(f"\nCategoria: {cat}")
    for p in prods:
        print(f"  - {p['nome']} (Estoque: {p['dados_estoque_preco'][0]} | Preço: R${p['dados_estoque_preco'][1]})")
```
