```python
#Exercício 19

palavra = input("Digite uma palavra: ").lower()
tamanho = len(palavra)
meio = tamanho // 2  # Usamos // para divisão inteira

# Se o resto for 0, o número é PAR
if tamanho % 2 == 0:
    idx1 = meio - 1
    idx2 = meio
    letra1 = palavra[idx1]
    letra2 = palavra[idx2]

    print(f"Quantidade de letras: {tamanho} (par)")
    print(f"Índices das letras centrais: {idx1} e {idx2}")
    print(f"Letras centrais: '{letra1}' e '{letra2}'")

    # Verificação de dígrafo (letras iguais no centro)
    if letra1 == letra2:
        print(f"As letras centrais formam o dígrafo '{letra1}{letra2}'")

# Se o resto for 1, o número é ÍMPAR
else:
    print(f"Quantidade de letras: {tamanho} (ímpar)")
    print(f"Índice da letra central: {meio}")
    print(f"Letra central: '{palavra[meio]}'")
```

```python
# Exercício 20

while True:
    texto = input("\nDigite um texto de exatos 5 caracteres (ou 'sair' para encerrar): ")

    if texto.lower() == 'sair':
        break

    if len(texto) != 5:
        print("Erro: O texto não possui exatos 5 caracteres. Tente novamente.")
        continue # Volta para o início do loop

    print("\n--- MENU DE OPÇÕES ---")
    print("1 - Retornar tudo em MINÚSCULO")
    print("2 - Retornar tudo em MAIÚSCULO")
    print("-1 - Sair")

    resposta = input("Escolha uma opção: ")

    if resposta == "-1":
        print("Saindo...")
        break
    elif resposta == '1':
        print(f"Resultado: {texto.lower()}")
    elif resposta == '2':
        print(f"Resultado: {texto.upper()}")
    else:
        print("Opção inválida!")
```

```python
#Exercicio 27
def cifra_de_cesar(texto, deslocamento):
    resultado = ""

    for char in texto:
        # Check if it is a standard uppercase letter (A-Z)
        if 'A' <= char <= 'Z':
            # Shift the letter and wrap around the alphabet
            novo_char = chr((ord(char) - ord('A') + deslocamento) % 26 + ord('A'))
            resultado += novo_char

        # Check if it is a standard lowercase letter (a-z)
        elif 'a' <= char <= 'z':
            # Shift the letter and wrap around the alphabet
            novo_char = chr((ord(char) - ord('a') + deslocamento) % 26 + ord('a'))
            resultado += novo_char

        # If it's a space, number, punctuation, or accented character, keep it as is
        else:
            resultado += char

    return resultado

# Input section matching the example
texto_entrada = input("Digite um texto: ")
coeficiente = int(input("Digite um coeficiente de deslocamento: "))

# Processing
texto_saida = cifra_de_cesar(texto_entrada, coeficiente)

# Output section matching the example
print(f"\nTexto criptografado: {texto_saida}")
```
