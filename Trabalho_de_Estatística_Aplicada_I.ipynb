```r
# UNIVERSIDADE FEDERAL DE SANTA CATARINA
# CENTRO DE CIÊNCIAS DA EDUCAÇÃO
# DEPARTAMENTO DE CIÊNCIAS DA INFORMAÇÃO
# CURSO DE CIÊNCIA DE DADOS
# Autor: Matheus Wander da Silva
# Estatística Aplicada I: Trabalho Sobre Variáveis Aleatórias Discretas

# ---------------------------------------------------------
# a) e b) Definição da Variável Aleatória X e Probabilidades
# ---------------------------------------------------------
# X = 0: Verde | X = 1: Amarelo | X = 2: Vermelho

estados <- c("Verde", "Amarelo", "Vermelho")
x <- c(0, 1, 2)
p <- c(0.40, 0.10, 0.50)

# Verificando a condição axiomática (A soma deve ser 1)
soma_p <- sum(p)

# Criando um Data Frame para visualizar a distribuição
distribuicao <- data.frame(Estado = estados, X = x, Probabilidade = p)

cat("--- Distribuição de Probabilidade ---\n")
print(distribuicao)
cat("Soma das Probabilidades:", soma_p, "\n\n")


# ---------------------------------------------------------
# c) Cálculo da Esperança Matemática E(X)
# ---------------------------------------------------------
# Fórmula: E(X) = Σ [x * P(X = x)]

esperanca_x <- sum(x * p)
cat("--- Medidas de Posição ---\n")
cat("Esperança Matemática E(X) =", esperanca_x, "\n\n")


# ---------------------------------------------------------
# d) Cálculo da Variância e Desvio Padrão de X
# ---------------------------------------------------------
# 1. Cálculo de E(X²)
esperanca_x2 <- sum((x^2) * p)

# 2. Cálculo da Variância Var(X) = E(X²) - [E(X)]²
variancia_x <- esperanca_x2 - (esperanca_x^2)

# 3. Cálculo do Desvio Padrão (σ) = √Var(X)
desvio_padrao <- sqrt(variancia_x)

cat("--- Medidas de Dispersão ---\n")
cat("E(X²) =", esperanca_x2, "\n")
cat("Variância Var(X) =", variancia_x, "\n")
cat("Desvio Padrão (σ) =", round(desvio_padrao, 3), "\n\n")


# ---------------------------------------------------------
# Visualização Gráfica (Bônus)
# ---------------------------------------------------------
# Gera um gráfico de barras com as cores dos semáforos
barplot(p,
        names.arg = paste(estados, "\n(X =", x, ")"),
#        main = "Distribuição de Probabilidades do Semáforo",
        xlab = "Estado do Semáforo (X)",
        ylab = "Probabilidade P(X = x)",
        col = c("#2ECC71", "#F1C40F", "#E74C3C"), # Cores verde, amarelo e vermelho
        ylim = c(0, 0.6),
        border = "black")
```
