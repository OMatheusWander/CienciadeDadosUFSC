```python
#Exercício 6

lista = [4, 7, 10, 3, 8, 11, 2]
lista_pares = []
lista_impares = []

print(f'Lista original: {lista}')

for valor in lista:
  if valor % 2 == 0:
    lista_pares.append(valor)
  else:
    lista_impares.append(valor)

print(f'Lista de pares: {lista_pares}')
print(f'Lista de impares: {lista_impares}')
print(f'Quantidade de pares: {len(lista_pares)}')
print(f'Quantidade de impares: {len(lista_impares)}')
```
