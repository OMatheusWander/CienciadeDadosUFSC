```python
# Criando a lista com 10 funcionários
funcionarios = [
    {"nome": "Ana Costa", "setor": "Tecnologia da Informação", "salario": 8500.00},
    {"nome": "Carlos Eduardo Silva", "setor": "Recursos Humanos", "salario": 5200.00},
    {"nome": "Mariana Oliveira", "setor": "Marketing", "salario": 6300.00},
    {"nome": "Roberto Almeida", "setor": "Vendas", "salario": 4800.00},
    {"nome": "Fernanda Lima", "setor": "Financeiro", "salario": 7100.00},
    {"nome": "João Pedro Santos", "setor": "Operações", "salario": 3900.00},
    {"nome": "Beatriz Souza", "setor": "Atendimento ao Cliente", "salario": 3500.00},
    {"nome": "Lucas Martins", "setor": "Logística", "salario": 4200.00},
    {"nome": "Camila Rocha", "setor": "Design", "salario": 5800.00},
    {"nome": "Ricardo Mendes", "setor": "Diretoria", "salario": 15000.00}
]

# Exibindo os dados formatados na tela
print("## Lista de Funcionários\n")
soma_salarios = 0
# CORREÇÃO 1: Mudar de [] (lista) para {} (dicionário)
funcionarios_agrupados = {}

for i, funcionario in enumerate(funcionarios, start=1):
    print(f"* **Funcionário {i}**")
    print(f"    * **Nome:** {funcionario['nome']}")
    print(f"    * **Setor:** {funcionario['setor']}")

    # Formatando o salário para o padrão brasileiro (R$ X.XXX,XX)
    salario_formatado = f"{funcionario['salario']:_.2f}".replace('.', ',').replace('_', '.')
    print(f"    * **Salário:** R$ {salario_formatado}")
    print() # Linha em branco para separar os funcionários

    soma_salarios += funcionario['salario']

# Percorrendo a lista para agrupar
for funcionario in funcionarios:
    setor = funcionario["setor"]

    # Se o setor ainda não existe no dicionário, criamos uma lista vazia para ele
    if setor not in funcionarios_agrupados:
        # CORREÇÃO 2: Usar a variável [setor] sem aspas, para criar a chave com o nome do departamento real
        funcionarios_agrupados[setor] = []

    # Adicionamos o nome do funcionário na lista do seu respectivo setor
    funcionarios_agrupados[setor].append(funcionario["nome"])

media_salarios = soma_salarios / len(funcionarios)

# Exibindo os resultados finais
print(f"Média salarial da empresa: R$ {media_salarios:.2f}\n")
print("Funcionários por setor:")

# Imprimindo o dicionário de forma mais bonita para visualizar melhor
for setor, nomes in funcionarios_agrupados.items():
    print(f"- {setor}: {nomes}")
```

```python
# Exemplo de lista de entrada
estudantes = [
    ["João", ["Matemática", "História", "Geografia"]],
    ["Maria", ["Biologia", "Química"]],
    ["Pedro", ["Matemática", "Física", "Química", "Biologia"]],
    ["Ana", ["História", "Geografia", "Artes"]]
]

# Inicialização de variáveis
maior_qtd = 0
estudante_recordista = ""
total_com_repeticao = 0
lista_disciplinas_unicas = set() # O 'set' remove duplicatas automaticamente
nomes_mais_de_3 = []

print("--- Relatório de Estudantes ---")

for item in estudantes:
    nome = item[0]
    disciplinas = item[1]
    qtd = len(disciplinas)

    # a) Imprimir nome e disciplinas
    print(f"Estudante: {nome} | Disciplinas: {', '.join(disciplinas)}")

    # b) Informar quantas disciplinas cada um cursa
    print(f"  -> Total: {qtd} disciplinas")

    # c) Lógica para maior estudante
    if qtd > maior_qtd:
        maior_qtd = qtd
        estudante_recordista = nome

    # d) Cálculos de total
    total_com_repeticao += qtd
    for disc in disciplinas:
        lista_disciplinas_unicas.add(disc) # Adiciona ao conjunto (ignora duplicadas)

    # e) Lista de nomes com 3 ou mais disciplinas
    if qtd >= 3:
        nomes_mais_de_3.append(nome)

# Exibição dos resultados finais
print("-" * 30)
print(f"Estudante com mais disciplinas: {estudante_recordista} ({maior_qtd} disciplinas)")
print(f"Total de disciplinas (com repetição): {total_com_repeticao}")
print(f"Total de disciplinas (sem repetição): {len(lista_disciplinas_unicas)}")
print(f"Estudantes com 3 ou mais disciplinas: {', '.join(nomes_mais_de_3)}")
```

```python
#Exercício 39

alunos_turma = (
    ("Ana", 7.5, 8.0),
    ("Bruno", 6.0, 5.5),
    ("Carla", 9.0, 8.5),
    ("Daniel", 4.5, 6.0),
    ("Eduarda", 8.0, 9.0),
    ("Felipe", 7.0, 7.5),
    ("Gabriela", 9.5, 10.0),
    ("Hugo", 5.0, 5.0),
    ("Isabela", 8.5, 9.0),
    ("João", 6.5, 7.0)
)

print("Estudantes:")
maior_media = soma_notas = acima_media = 0
Estudante_maior_média = ""
for aluno, nota_1, nota_2 in alunos_turma:
  media = (nota_1 + nota_2)/2
  if media > maior_media:
    maior_media = media
    Estudante_maior_média = aluno
  soma_notas += nota_1 + nota_2
  print(f'- {aluno}. Notas: {nota_1}, {nota_2}. Média: {media}')

print(f"Estudante que possui a maior média: {Estudante_maior_média}")
print(f"Média geral da turma: {soma_notas/20}")

for aluno, nota_1, nota_2 in alunos_turma:
  media = (nota_1 + nota_2)/2
  if media > (soma_notas/20):
    acima_media += 1

print(f"Total de alunos que ficaram acima da média geral: {acima_media}")
```

```python
#Exercício 40 - REVISAR

funcionarios = [
    ['Fulano da Silva', 1000.00],
    ['Beltrano de Souza', 2230.00],
    ['Sicrano dos Santos', 2500.10],
    ['Maria das Couves', 3000.00],
    ['João Ninguém', 4000.00],
    ['Maria Chiquinha', 2230.00],
    ['João Valentão', 1472.00],
    ['Nonô Sem Dente', 27050.20],
    ['Robernélson da Luz', 3102.05],
    ['José Maria João', 1000.35],
    ['Agrícola Beterraba Areia Leão', 1500.00],
    ['Kenquem que Ninguém Quer', 800.00],
    ['Lindolfa Celidônia Calafange de Tefé', 25000.00],
    ['Simplício Simplório da Simplicidade Simples', 10437.22],
    ['Anônimo João Torquato', 7200.00],
    ['Kayhh-qui Caôlho e Corcunda', 5427.32]
]

# Processamento
total_salarios = 0
contagem_maria = 0
contagem_joao = 0

# Impressão dos salários em blocos de 5
for i in range(0, len(funcionarios), 5):
    print(75 * "#")
    print(f"Salários {i+1} a {min(i+5, len(funcionarios))}".center(75))
    print(75 * "#")

    for j in range(i, min(i+5, len(funcionarios))):
        nome, salario = funcionarios[j]
        # Conta caracteres (ignorando espaços se necessário, mas na imagem parece contar tudo)
        num_letras = len(nome.replace(" ", ""))
        # Nota: A imagem conta caracteres totais, incluindo espaços:
        num_letras = len(nome)

        print(f"Salário {j+1} ({nome}: {num_letras} letras): R$ {salario:.2f}")

        total_salarios += salario
        contagem_maria += nome.lower().count('maria')
        contagem_joao += nome.lower().count('joão')

# Cálculos finais
media = total_salarios / len(funcionarios)
parte_inteira = int(media)

print(75* "#")
print(f"Total somado dos salários dos funcionários: R$ {total_salarios:.2f}")
print(f"Média salarial da empresa: R$ {media:.2f}")
print(f"A parte inteira da média {parte_inteira} não é divisível por 37.")
print(f"A parte inteira da média {parte_inteira} não é divisível por 22.")
print(f"A parte inteira da média {parte_inteira} não é divisível por 10.")
print(75 * "#")
print(f"Ocorrências da palavra Maria: {contagem_maria} (é um número {'ímpar' if contagem_maria % 2 != 0 else 'par'}).")
print(f"Ocorrências da palavra João: {contagem_joao} (é um número {'ímpar' if contagem_joao % 2 != 0 else 'par'}).")
print(75 * "#")
```
